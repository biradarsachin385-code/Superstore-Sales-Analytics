```python
subcategory = df.groupby('Sub-Category')[['Sales', 'Profit']].sum()

subcategory['Profit Margin %'] = (
    subcategory['Profit'] / subcategory['Sales']
) * 100
subcategory.sort_values('Profit Margin %')
```

```python
import matplotlib.pyplot as plt

margin = subcategory.sort_values('Profit Margin %')

plt.figure(figsize=(10, 7))
plt.barh(margin.index, margin['Profit Margin %'])
plt.xlabel('Profit Margin (%)')
plt.ylabel('Sub-Category')
plt.title('Profit Margin by Sub-Category')
plt.axvline(0, linewidth =1)
plt.tight_layout()
plt.show()
```

```python
region_analysis = df.groupby('Region')[['Sales', 'Profit']].sum()

region_analysis['Profit Margin %'] = (
    region_analysis['Profit'] / region_analysis['Sales']
) * 100

region_analysis.sort_values('Profit', ascending=False)
```

```python
region_category = df.groupby(
    ['Region', 'Category']
)[['Sales', 'Profit']].sum()

region_category
```

```python
region_category['Profit Margin %'] = (
    region_category['Profit'] / region_category['Sales']
) * 100

region_category.sort_values('Profit Margin %')
```
